```python
# Import Libraries 

import pandas as pd
import numpy as np
import matplotlib.pyplot as plot
import seaborn as sns

# Graph show in notebook

%matplotlib inline

# style

sns.set_style("whitegrid")
```

```python
# data set load
df = pd.read_csv("../data/hospital_patient_records.csv")
```

```python
# Apply  EDA (Exploratory Data Analysis)

# First 5 rows
df.head()

# Last 5 rows
df.tail()

# shape
df.shape

# columns names
df.columns

# data types
df.dtypes

# data information
df.info()

# Statistical summary
df.describe()

# missing values
df.isnull().sum()

# Duplicate Rows
df.duplicated().sum()

# Unique values
df.nunique()

```

```python
# Data Preprocessing

# step(1): missing values Handle 

df.isnull().sum()
```

```python
# fill values::

df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])

df["BMI"] = df["BMI"].fillna(df["BMI"].median())

df["HbA1c"] = df["HbA1c"].fillna(df["HbA1c"].median())

df["Blood_Glucose"] = df["Blood_Glucose"].fillna(df["Blood_Glucose"].median())

df["Hemoglobin"] = df["Hemoglobin"].fillna(df["Hemoglobin"].median())

```

```python
# Step 2: Duplicate Remove

df = df.drop_duplicates()

df.duplicated().sum()
```

```python
df.to_csv("../data/hospital_patient_records_cleaned.csv", index=False)
print("✅ Cleaned dataset saved successfully!")
```
